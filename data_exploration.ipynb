```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r'C:\Users\mechz\Documents\Jobs_Analysis\job_postings.csv')
display(df)

columns = df.columns.tolist()
country_columns = [column for column in columns if 'job_country' in column]
via_columns = [column for column in columns if 'job_via' in column]
skill_columns = [column for column in columns if 'skill' in column]
schedule_columns = ['Contractor', 'Full-time', 'Internship', 'Part-time', 'Tempwork']
job_title_columns = [column for column in columns if 'job_title_short' in column]

print('Count for country columns')
for column in country_columns:
    print(df[column].sum())

print('Count for via columns')
for column in via_columns:
    print(df[column].sum())

print('Count for skill columns')
for column in skill_columns:
    print(df[column].sum())

print('Count for schedule columns')
for column in schedule_columns:
    print(df[column].sum())

print('Count for job title columns')
for column in job_title_columns:
    print(df[column].sum())
```
